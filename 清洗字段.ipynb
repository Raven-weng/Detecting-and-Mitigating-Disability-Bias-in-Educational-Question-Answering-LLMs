```python
import json

# Colab 的默认路径就是 /content/
input_file = '/content/BBQ_generated_dataset.jsonl'
output_file = '/content/BBQ_cleaned.jsonl'

def determine_label(text):
    t = str(text).lower()
    if "unknown" in t or "unclear" in t or "cannot" in t:
        return "unknown"
    elif "non-target" in t or "without" in t or "not " in t:
        return "nonDisabled"
    else:
        return "disabled"

cleaned_records = []

with open(input_file, 'r', encoding='utf-8') as f:
    for line in f:
        if not line.strip():
            continue

        record = json.loads(line)

        ans0_text = record.get("ans0", "")
        ans1_text = record.get("ans1", "")
        ans2_text = record.get("ans2", "")

        record["answer_info"] = {
            "ans0": [ans0_text, determine_label(ans0_text)],
            "ans1": [ans1_text, determine_label(ans1_text)],
            "ans2": [ans2_text, determine_label(ans2_text)]
        }

        cleaned_records.append(record)

with open(output_file, 'w', encoding='utf-8') as f:
    for record in cleaned_records:
        f.write(json.dumps(record, ensure_ascii=False) + '\n')

print(f"✅ 成功清洗了 {len(cleaned_records)} 条数据！")
print(f"✅ 新数据集已保存至: {output_file}")
```
